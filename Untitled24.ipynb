```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. PARAMETRELER (1860-1914 Kalibrasyonu) ---
start_year = 1860
end_year = 1914
T = end_year - start_year + 1
N_sim = 1000

# Şok Parametreleri
theta_xi = 0.65
sigma_eps = 0.25
xi_star = 1.15
xi_0 = 0.2

# Devlet Baskısı ve Bürokrasi Ağırlıkları (Denklem 8)
mu_1, mu_2, mu_3 = 0.3, 0.2, 0.4  # Muhtar, Damga Pulu, Jandarma
alpha_bureaucracy = 1.2           # Bürokrasi Esneklik Parametresi (Denklem 12)

# Logit Tercih ve Pazar Parametreleri
I_param = 2.0   # Tercih Yoğunluğu
P_price = 2.0   # Ürün Satış Fiyatı
Lambda_jandarma = 0.3 # Yakalanma Etkinliği

# Başlangıç Değerleri
S_0 = 0.8       # Başlangıç Kurumsal Kalite Endeksi
K_0 = 100.0     # Başlangıç Fiziksel Sermaye
gamma = 0.5     # TFV Esnekliği
A_base = 1.0    # Taban Verimlilik

# Depolama Matrisleri
xi_path = np.zeros((T, N_sim))
prob_coop_path = np.zeros((T, N_sim))
S_path = np.zeros((T, N_sim))

# --- 2. MONTE CARLO - ADIM 1 ve ADIM 2 BİRLEŞİK DÖNGÜ ---
np.random.seed(1860)

for sim in range(N_sim):
    xi = xi_0
    S = S_0
    K = K_0

    for t in range(T):
        # --- Adım 1: Stokastik Şok Süreci (Denklem 9) ---
        eps_xi = np.random.normal(0, sigma_eps)
        xi = theta_xi * xi + eps_xi
        xi = max(0.0, xi)
        xi_path[t, sim] = xi

        # --- Adım 2.1: Bürokrasi ve Devlet Baskısı Maliyeti (Denklem 8 & 12) ---
        # Sabit veya şoka bağlı bürokratik kalemler (omega 1, 2, 3 varsayılan temsili değerler)
        omega_1, omega_2, omega_3 = 0.4, 0.3, 0.5
        rho_t = mu_1 * omega_1 + mu_2 * omega_2 + mu_3 * omega_3

        # Tezkere ve Belgelendirme Bürokrasi Endeksi (Denklem 12)
        Gamma_t = (xi ** alpha_bureaucracy) * (rho_t ** alpha_bureaucracy)

        # --- Adım 2.2: Çıktı ve Strateji Faydaları (Denklem 1, 17, 18) ---
        Y_t = A_base * (S ** gamma) * K  # Çıktı

        # İşbirliği Net Faydası (Denklem 17)
        V_coop = P_price * Y_t - rho_t

        # Fırsatçılık/Hırsızlık Net Faydası (Denklem 18)
        # Şok arttıkça çalınan sürünün değeri (Omega_t) ve fırsatçılık cazibesi artar, ancak yakalanma riski (q*L) cezalandırır
        Omega_t = 5.0 * xi # Çalınan sürünün potansiyel değeri
        q_penalty = Lambda_jandarma * rho_t * 2.0 # Yakalanma ve ceza maliyeti
        V_opp = P_price * Y_t + Omega_t - rho_t - q_penalty

        # --- Adım 2.3: Logit (Olasılıklı Tercih) Fonksiyonu (Denklem 21) ---
        # Taşrada rasyonel sınırlı bilgi altında işbirliğini tercih etme oranı
        exp_coop = np.exp(I_param * V_coop)
        exp_opp = np.exp(I_param * V_opp)
        prob_coop = exp_coop / (exp_coop + exp_opp)

        prob_coop_path[t, sim] = prob_coop

        # --- Adım 2.4: Kurumsal Kalite Güncellemesi (Önizleme - Adım 3 için) ---
        vartheta_fb = 0.3
        beta_deg = 0.4
        berg_eff = 1.0 if xi <= xi_star else 0.0

        delta_S = vartheta_fb * prob_coop - beta_deg * (1.0 - prob_coop) * xi * (1 - berg_eff)
        S = max(0.1, min(2.0, S + delta_S))
        S_path[t, sim] = S

# --- 3. GÖRSELLEŞTİRME: İŞBİRLİĞİ ORANI PATİKASI ---
years = np.arange(start_year, end_year + 1)
mean_coop = np.mean(prob_coop_path, axis=1)
p5_coop = np.percentile(prob_coop_path, 5, axis=1)
p95_coop = np.percentile(prob_coop_path, 95, axis=1)

plt.figure(figsize=(10, 5))
plt.plot(years, mean_coop, label='Ortalama İşbirliği Olasılığı Prob(Strategy 1)', color='midnightblue', linewidth=2)
plt.fill_between(years, p5_coop, p95_coop, color='blue', alpha=0.15, label='%90 Güven Aralığı')
plt.title('1860-1914 Osmanlı Kırsalı: Logit Tabanlı Toplumsal İşbirliği Olasılığı')
plt.xlabel('Yıllar')
plt.ylabel('İşbirliği Oranı [0, 1]')
plt.legend(loc='lower left')
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. PARAMETRELER (Kararlı ve Dalgalı Denge Seti) ---
start_year = 1860
end_year = 1914
years = np.arange(start_year, end_year + 1)
T = len(years)
N_sim = 1000

# Şok ve Eşik Parametreleri (AR1)
theta_xi = 0.60
sigma_eps = 0.20
xi_star = 1.15
xi_0 = 0.2

# Sermaye ve Amortisman Parametreleri
delta_K = 0.05
delta_animal = 0.07
A_base = 1.0
gamma = 0.25

# Devlet ve Bürokrasi Parametreleri
mu_1, mu_2, mu_3 = 0.2, 0.15, 0.3
alpha_bureaucracy = 1.0
I_param = 1.5
P_price = 2.0
Lambda_jandarma = 0.4

# Kurumsal Geri Besleme ve Yozlaşma Katsayıları
vartheta_fb = 0.08
beta_deg = 0.12

# Başlangıç Değerleri
S_0 = 1.0
K_0 = 80.0
K_animal_0 = 60.0

# Depolama Matrisleri
S_path = np.zeros((T, N_sim))
K_animal_path = np.zeros((T, N_sim))

# --- 2. MONTE CARLO DÖNGÜSÜ ---
np.random.seed(1860)

for sim in range(N_sim):
    xi = xi_0
    S = S_0
    K = K_0
    K_animal = K_animal_0

    for t in range(T):
        # Stokastik Şok Süreci
        eps_xi = np.random.normal(0, sigma_eps)
        xi = theta_xi * xi + eps_xi
        xi = max(0.0, xi)

        berg_eff = 1.0 if xi <= xi_star else 0.0

        # Bürokrasi ve Fayda
        rho_t = mu_1 * 0.3 + mu_2 * 0.2 + mu_3 * 0.4
        Y_t = A_base * (S ** gamma) * K

        V_coop = P_price * Y_t - rho_t
        Omega_t = 1.5 * xi
        q_penalty = Lambda_jandarma * rho_t * 1.0
        V_opp = P_price * Y_t + Omega_t - rho_t - q_penalty

        exp_coop = np.exp(I_param * V_coop)
        exp_opp = np.exp(I_param * V_opp)
        prob_coop = exp_coop / (exp_coop + exp_opp)

        # Kurumsal Kalite Güncellemesi (Doğal Salınımlı)
        # Kurumların aşırı büyümesini engelleyen içsel bir yaylanma (reversion) eklendi
        delta_S = vartheta_fb * prob_coop - beta_deg * (1.0 - prob_coop) * xi * (1 - berg_eff) - 0.03 * (S - 1.0)
        S = S + delta_S
        S = max(0.5, S) # Sadece aşırı düşüşü engelleyen taban sınırı
        S_path[t, sim] = S

        # Fiziksel Sermaye
        I_investment = 0.12 * Y_t
        K = (1.0 - delta_K) * K + I_investment - 0.02 * K # Azalan getiri/aşınma baskısı

        # Canlı Hayvan Sermayesi (Doğal Dalgalanma)
        phi_t = (1.0 - prob_coop) * xi * 0.5
        I_animal = 0.07 * Y_t
        K_animal = (1.0 - delta_animal) * K_animal + I_animal - phi_t
        K_animal = max(20.0, K_animal) # Tabana çakılmayı önleyen yumuşak alt sınır
        K_animal_path[t, sim] = K_animal

# --- 3. GÖRSELLEŞTİRME (1860-1914 Dalgalı ve Kararlı Patika) ---
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(years, np.mean(S_path, axis=1), color='darkred', linewidth=2)
plt.title('1860-1914 Kurumsal Kalite ve Güvenlik ($S_t$)')
plt.xlabel('Yıllar')
plt.ylabel('Endeks Değeri')
plt.xlim(1860, 1914)
plt.grid(True, linestyle=':', alpha=0.6)

plt.subplot(1, 2, 2)
plt.plot(years, np.mean(K_animal_path, axis=1), color='darkgreen', linewidth=2)
plt.title('1860-1914 Canlı Hayvan Sermayesi ($K^{animal}_t$)')
plt.xlabel('Yıllar')
plt.ylabel('Stok Miktarı')
plt.xlim(1860, 1914)
plt.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
